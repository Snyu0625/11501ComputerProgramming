問題：撰寫一Python程式，為使用者輸入年份（可能是西元或民國）找尋其相對應的十二生肖。十二生肖是以十二年為一週期，每一年分別以一種動物來表示猴（monkey）、雞（rooster）、狗(dog)、豬(pig)、鼠(rat)、牛(ox)、虎(tiger)、兔(rabbit)、龍(dragon)、蛇（snake）、馬(horse)，以及羊(sheep) 。

````python
def get_chinese_zodiac():
    # 依照題目指定的順序建立十二生肖串列（索引 0 對應餘數 0，以此類推）
    zodiac_animals = ['猴', '雞', '狗', '豬', '鼠', '牛', '虎', '兔', '龍', '蛇', '馬', '羊']
    
    print("=== 十二生肖查詢系統 ===")
    print("請選擇您要輸入的年份類型：")
    print("1. 西元年份 (例如: 2026)")
    print("2. 民國年份 (例如: 115)")
    
    choice = input("請輸入選項代號 (1 或 2)：").strip()
    
    if choice == '1':
        try:
            year = int(input("請輸入西元年份："))
            west_year = year
        except ValueError:
            print("錯誤：請輸入有效的數字年份！")
            return
    elif choice == '2':
        try:
            roc_year = int(input("請輸入民國年份："))
            west_year = roc_year + 1911  # 民國年份轉為西元年
        except ValueError:
            print("錯誤：請輸入有效的數字年份！")
            return
    else:
        print("錯誤：無效的選項。")
        return

    # 計算西元年份除以 12 的餘數，對應生肖串列
    index = west_year % 12
    animal = zodiac_animals[index]
    
    # 輸出結果
    print("\n" + "="*30)
    print(f"西元 {west_year} 年（民國 {west_year - 1911} 年）")
    print(f"對應的生肖為：【 {animal} 】")
    print("="*30)

if __name__ == "__main__":
    get_chinese_zodiac()
````
