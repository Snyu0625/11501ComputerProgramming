使用者輸入：請輸入正整數$n$，其中$1\le n\le 9$，輸出二進位$nn$乘法表。格式如如Tranclass的圖片Exercise20261008.png，以$n=5$為例。                                                                                                         
                                                                                                            
1. 只能使用while。                                                                                            
3. 字串格式化只能使用 f-string 對齊。                                                                       

```python
# 取得使用者輸入的正整數 n (1 <= n <= 9)
n = int(input("請輸入正整數 n (1<=n<=9): "))

# 使用巢狀 for 迴圈逐行、逐列印出乘法表
for i in range(1, n + 1):
  row_list = []
  for j in range(1, n + 1):
    # 計算相乘結果
    product = i * j

    # 轉換成二進位字串並使用 f-string 補零對齊
    # i, j 轉為 3位二進位；product 轉為 6位二進位
    item = f"{i:03b}x{j:03b}={product:06b}"
    row_list.append(item)

  # 將每一行的項目用空格串接並輸出
  print("   ".join(row_list))
```

```python
# 取得使用者輸入的正整數 n (1 <= n <= 9)
n = int(input("請輸入正整數 n (1<=n<=9): "))

# 初始化外層 while 迴圈變數
i = 1
while i <= n:
  row_list = []
  j = 1
  # 初始化內層 while 迴圈變數
  while j <= n:
    product = i * j

    # 轉換成二進位字串，使用 f-string 補零對齊（乘號使用圖片中的 '×'）
    item = f"{i:03b}×{j:03b}={product:06b}"
    row_list.append(item)
    j += 1

  # 將每一行的項目用空格串接並輸出
  print("   ".join(row_list))
  i += 1
```
