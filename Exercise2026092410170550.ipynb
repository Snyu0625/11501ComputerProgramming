## 第三部分：CPython 如何實作這些型別？

除了知道 Python 使用哪些型別表示數學物件之外，

> Python 物件在 CPython 中是如何實作的？

請查閱資料，完成下表。

| 數學概念 | Python 型別 | CPython 實作 | 主要特點 |
|----------|------------|-------------|----------|
| 自然數 | `int` | PyLongObject | 在Python中自然數與整數統一處理；支援任意精度長整數，長度可動態擴充而不易溢位 |
| 整數 | `int` | PyLongObject | 使用陣列儲存大數；小整數會被快取重用 |
| 實數 | `float` | PyFloatObject | 底層直接封裝C語言的double |

| 集合 | `set` | PySetObject | 基於雜湊表實作；元素必須可雜湊，提供平均O(1)的尋找與插入時間複雜度 |
| 命題真假 | `bool` | PyLongObject | bool是int的子類別，全區域僅有單例物件Py_True與Py_False |
| 串列 | `list` | PyListObject | 底層為動態指標陣列，支援隨機存取O(1)，記憶體預分配以達到 O(1)的append操作 |

## 第五部分：Python 整數與浮點數

執行下列程式：

```python
print(10**100)
print(10**1000)
```

回答：

1. 為什麼 Python 能表示如此大的整數？因為Python不使用 CPU 的硬體原生整數來直接代表 Python 的 int
2. Python 的 `int` 是否具有固定的位元數限制？否
3. 什麼是 Arbitrary Precision（任意精度）？數值可表達的位數與精度不受硬體暫存器大小的限制，僅受限於可用記憶體的大小
4. CPython 使用什麼物件來實作 `int`？PyLongObject

## 第六部分：AI 模型中的數值表示

人工智慧模型需要儲存大量參數（Parameters）與權重（Weights）。

請查閱資料並回答：

1. AI 模型中的權重通常使用什麼型別儲存？低精度的浮點數或量化後的整數儲存
2. 什麼是 FP64？佔64位元。精度與範圍極高
3. 什麼是 FP32？佔32位元。深度學習傳統的標準格式
4. 什麼是 FP16？佔16位元。記憶體佔用為FP32的一半，計算速度大幅提升
5. 什麼是 BF16？佔 16 位元。直接繼承了FP32的8位元指數，具有與FP32完全相同的數值動態範圍
6. 為什麼大型 AI 模型不直接使用 Python 的 `int` 作為權重？Python int 是物件而非原生數值，記憶體開銷過大、缺少小數連續表示能力、缺少 GPU/NPU 硬體加速支援、記憶體佈局不連續

---

### 思考題二

比較：

```python
int
```

與

```python
float
```

回答：

1. 為什麼 `int` 幾乎沒有固定上限？int的大小只受限於電腦實體記憶體的大小，沒有固定的位元數或數值上限
2. 為什麼 `float` 卻存在範圍限制？指數部分被限制在11個位元內，能表示的最大數量級約為 $1.79 \times 10^{308}$。就會產生溢位，直接轉為inf
3. 為什麼 `float` 可能產生誤差？二進位無法精確表示部分十進位小數、有限位數帶來的截斷、誤差累積

### 思考題三

比較下列兩種設計理念：

#### Python 的整數

```text
Arbitrary Precision
正確性優先
```

#### AI 模型常用的 FP16、BF16

```text
Limited Precision
效率優先
```

請說明：

1. 兩種設計理念的差異。Python正確性優先、動態配給、無理論上限、軟體層級控制；AI效率優先、固定分配、有嚴格邊界、硬體級原生加速
2. 各自的優點與缺點。Python優點:絕對精確、程式寫法直覺。缺點:記憶體與計算開銷極大、無法平行加速；AI優點:計算速度極快、記憶體佔用極低。缺點:存在捨入誤差、數值不穩定風險
3. 為什麼大型語言模型通常選擇較低精度的表示方式？容錯性與魯棒性、顯存瓶頸、計算與傳輸速度
4. 如果所有 AI 模型都改用 FP64，可能會產生哪些問題？顯存需求爆發、計算速度斷崖式下跌、電力與能源嚴重浪費
